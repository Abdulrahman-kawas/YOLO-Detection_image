```python
import cv2
import numpy as np
from PIL import Image
from google.colab.patches import cv2_imshow

def show_image(image_path, boxes, model, conf1=0.4):
    image = Image.open(image_path)
    image = np.array(image)

    for box in boxes:
        confidence = float(box.conf)
        if confidence < conf1:
            continue

        cls_id = int(box.cls)
        class_name = model.names[cls_id]

        x1, y1, x2, y2 = box.xyxy[0].tolist()

        cv2.rectangle(
            image,
            (int(x1), int(y1)),
            (int(x2), int(y2)),
            (0, 255, 0),
            2
        )

        label = f"{class_name} {confidence:.2f}"
        cv2.putText(
            image,
            label,
            (int(x1), int(y1) - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            2,
            (0, 255, 0),
            4
        )

    cv2_imshow(image)
```

```python
show_image(image_path , boxes, model)
```
